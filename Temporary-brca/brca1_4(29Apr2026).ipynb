```python
# =========================================================
# STEP 2 — Download Multiple GEO Datasets
# Datasets:
# GSE10810
# GSE42568
# GSE65194
# GSE45827
# GSE209998 baaad eta (RNA seq)
# =========================================================

!pip install GEOparse -q

import GEOparse
import os

# Create GEO folder
os.makedirs("GEO_Datasets", exist_ok=True)

# Dataset IDs
geo_ids = [
    "GSE10810",
    "GSE42568",
    "GSE65194",
    "GSE45827"
    #"GSE209998"   # NEW RNA-seq dataset
]

# Dictionary to store datasets
gse_data = {}

# Download datasets
for geo_id in geo_ids:

    print("="*60)
    print(f"Downloading {geo_id} ...")

    try:
        gse = GEOparse.get_GEO(
            geo=geo_id,
            destdir="GEO_Datasets"
        )

        gse_data[geo_id] = gse

        print(f"{geo_id} Loaded Successfully ✅")
        print("Title:", gse.metadata['title'][0])
        print("Platforms:", list(gse.gpls.keys()))
        print("Samples:", len(gse.gsms))

    except Exception as e:
        print(f"Failed to load {geo_id} ❌")
        print("Error:", e)

print("\nAll datasets processed ✅")
```

```python
# =========================================================
# STEP 2.01 — Show Dataset Dimensions (Rows & Columns)
# =========================================================

import pandas as pd

dataset_shapes = []

for geo_id, gse in gse_data.items():

    print("="*70)
    print(f"Dataset: {geo_id}")

    # Take first sample to estimate dimensions
    first_sample = list(gse.gsms.keys())[0]

    # Expression table
    sample_table = gse.gsms[first_sample].table

    rows, cols = sample_table.shape

    print(f"Sample Table Shape: {rows} rows × {cols} columns")

    print("\nColumn Names:")
    print(sample_table.columns.tolist())

    dataset_shapes.append({
        "Dataset": geo_id,
        "Rows": rows,
        "Columns": cols,
        "Samples": len(gse.gsms)
    })

# Summary Table
summary_df = pd.DataFrame(dataset_shapes)

print("\n\n================ DATASET SUMMARY ================\n")
display(summary_df)
```

```python
# =========================================================
# STEP 3 — Build Expression Matrices for ALL GEO Datasets
# Research-Grade Version
# =========================================================

import pandas as pd
import numpy as np
import gc

# Store all expression matrices
expression_matrices = {}

# Loop through each GEO dataset
for geo_id, gse in gse_data.items():

    print("="*70)
    print(f"Processing {geo_id} ...")

    sample_tables = []

    # Extract all GSM samples
    for gsm_name, gsm in gse.gsms.items():

        try:
            # Extract ID_REF and VALUE
            df = gsm.table[['ID_REF', 'VALUE']].copy()

            # Rename columns
            df.columns = ['Gene', gsm_name]

            # Convert expression values to numeric
            df[gsm_name] = pd.to_numeric(
                df[gsm_name],
                errors='coerce'
            )

            # Remove duplicate probes
            df = df.drop_duplicates(subset='Gene')

            sample_tables.append(df)

        except Exception as e:
            print(f"Skipped {gsm_name}: {e}")

    # Merge all samples
    expr = sample_tables[0]

    for df in sample_tables[1:]:
        expr = expr.merge(df, on='Gene', how='inner')

    # Set gene names as index
    expr = expr.set_index('Gene')

    # Remove genes with too many missing values
    expr = expr.dropna(axis=0, thresh=int(expr.shape[1] * 0.8))

    # Fill remaining NA with median
    expr = expr.apply(
        lambda x: x.fillna(x.median()),
        axis=1
    )

    # Save matrix
    expression_matrices[geo_id] = expr

    # Print summary
    print(f"{geo_id} Expression Matrix Ready ✅")
    print("Shape:", expr.shape)

    # Show first rows
    display(expr.iloc[:5, :5])

    # Memory cleanup
    gc.collect()

print("\nALL Expression Matrices Built Successfully ✅")
```

```python
# =========================================================
# STEP 3.01 — Probe Annotation (Optimized Local Version)
# Convert Probe IDs to Gene Symbols using local GPL tables
# =========================================================

import pandas as pd

# Store annotated datasets
annotated_matrices = {}

for geo_id, expr in expression_matrices.items():

    print("="*70)
    print(f"Annotating {geo_id} ...")

    # Get the platform table from the already loaded gse_data
    gse = gse_data[geo_id]
    platform = list(gse.gpls.values())[0]
    annot = platform.table.copy()

    # Identify columns for ID and Gene Symbol
    # Most Affymetrix platforms use 'ID' and 'Gene Symbol'
    # We normalize them here
    id_col = 'ID'
    symbol_col = 'Gene Symbol' if 'Gene Symbol' in annot.columns else 'Symbol'

    if symbol_col not in annot.columns:
        # Fallback: find any column containing 'symbol'
        potential_cols = [c for c in annot.columns if 'symbol' in c.lower()]
        if potential_cols:
            symbol_col = potential_cols[0]

    # Clean annotation table
    annot = annot[[id_col, symbol_col]].dropna()
    annot.columns = ['Probe', 'GeneSymbol']

    # Handle cases where multiple symbols are separated by '///'
    annot['GeneSymbol'] = annot['GeneSymbol'].astype(str).str.split('///').str[0].str.strip()

    # Merge with expression matrix
    expr_reset = expr.reset_index()
    expr_reset.columns.values[0] = "Probe"

    merged = expr_reset.merge(
        annot,
        on='Probe',
        how='inner'
    )

    # Remove empty or NaN symbols
    merged = merged[merged['GeneSymbol'] != ""]
    merged = merged[merged['GeneSymbol'] != "nan"]

    # Save
    annotated_matrices[geo_id] = merged

    print(f"{geo_id} Annotation Complete ✅")
    print("Shape:", merged.shape)
    display(merged.head())

print("\nAll datasets annotated successfully using local metadata ✅")
```

```python
# =========================================================
# STEP 3.02 — Probe Aggregation
# Multiple probes -> single gene
# Using mean expression
# =========================================================

aggregated_matrices = {}

for geo_id, df in annotated_matrices.items():

    print("="*70)
    print(f"Aggregating {geo_id} ...")

    # Remove probe column
    expr_cols = [
        col for col in df.columns
        if col not in ['Probe', 'GeneSymbol']
    ]

    # Group by Gene Symbol
    aggregated = df.groupby('GeneSymbol')[expr_cols].mean()

    # Remove duplicated index
    aggregated = aggregated[~aggregated.index.duplicated()]

    # Save
    aggregated_matrices[geo_id] = aggregated

    print(f"{geo_id} Aggregation Complete ✅")
    print("Final Shape:", aggregated.shape)

    display(aggregated.iloc[:5, :5])

print("\nAll datasets aggregated successfully ✅")
```

```python
# =========================================================
# STEP 3.03 — Filter Low-Expression / Low-Variance Genes / PERFECT Filtering......
# Research-Grade Preprocessing
# =========================================================

import numpy as np

filtered_matrices = {}

for geo_id, expr in aggregated_matrices.items():

    print("="*70)
    print(f"Filtering {geo_id} ...")

    # Remove genes with too many missing values
    expr = expr.dropna(axis=0, thresh=int(expr.shape[1] * 0.8))

    # Fill remaining missing values with gene median
    expr = expr.apply(
        lambda x: x.fillna(x.median()),
        axis=1
    )

    # Remove low variance genes
    gene_variance = expr.var(axis=1)

    variance_threshold = np.percentile(
        gene_variance,
        25
    )

    expr_filtered = expr[
        gene_variance > variance_threshold
    ]

    # Remove extremely low-expression genes
    expr_filtered = expr_filtered[
        expr_filtered.mean(axis=1) > 1
    ]

    filtered_matrices[geo_id] = expr_filtered

    print(f"{geo_id} Filtering Complete ✅")
    print("Original Shape:", expr.shape)
    print("Filtered Shape:", expr_filtered.shape)

    display(expr_filtered.iloc[:5, :5])

print("\nAll datasets filtered successfully ✅")
```

```python
# =========================================================
# STEP 3.04 — Log2 Transformation
# Stabilize expression distribution
# =========================================================

import numpy as np

log2_matrices = {}

for geo_id, expr in filtered_matrices.items():

    print("="*70)
    print(f"Checking Log2 Transformation for {geo_id} ...")

    # Detect whether already log-transformed
    max_val = expr.max().max()

    print("Maximum Expression Value:", max_val)

    # If values are too large -> apply log2
    if max_val > 50:

        expr_log2 = np.log2(expr + 1)

        print("Log2 Transformation Applied ✅")

    else:

        expr_log2 = expr.copy()

        print("Already Log-Transformed ✅")

    log2_matrices[geo_id] = expr_log2

    print("Final Shape:", expr_log2.shape)

    display(expr_log2.iloc[:5, :5])

print("\nAll datasets processed successfully ✅")
```

```python
# =========================================================
# STEP 4 — Extract Sample Metadata and Group Labels
# Research-Grade Version
# =========================================================

import pandas as pd

# Store metadata for all datasets
metadata_dict = {}

for geo_id, gse in gse_data.items():

    print("="*70)
    print(f"Processing Metadata: {geo_id}")

    meta = []

    for gsm_name, gsm in gse.gsms.items():

        # Collect metadata text
        title = " ".join(
            gsm.metadata.get("title", [""])
        ).lower()

        source = " ".join(
            gsm.metadata.get("source_name_ch1", [""])
        ).lower()

        characteristics = " ".join(
            gsm.metadata.get("characteristics_ch1", [""])
        ).lower()

        text = f"{title} {source} {characteristics}"

        # Default
        group = "Unknown"
        subtype = "Unknown"

        # -------- Cancer vs Normal --------
        normal_keywords = [
            "normal",
            "healthy",
            "control",
            "adjacent normal",
            "reduction mammoplasty"
        ]

        cancer_keywords = [
            "tumor",
            "cancer",
            "carcinoma",
            "breast cancer",
            "tumour"
        ]

        if any(k in text for k in normal_keywords):
            group = "Normal"

        elif any(k in text for k in cancer_keywords):
            group = "Cancer"

        # -------- Molecular Subtypes --------
        if "triple negative" in text or "tnbc" in text:
            subtype = "TNBC"

        elif "her2" in text:
            subtype = "HER2+"

        elif "luminal a" in text:
            subtype = "Luminal A"

        elif "luminal b" in text:
            subtype = "Luminal B"

        # Save row
        meta.append([
            gsm_name,
            group,
            subtype,
            title
        ])

    # Build dataframe
    meta_df = pd.DataFrame(
        meta,
        columns=[
            "Sample",
            "Group",
            "Subtype",
            "Title"
        ]
    )

    metadata_dict[geo_id] = meta_df

    # Print summary
    print("\nGroup Counts:")
    print(meta_df["Group"].value_counts())

    print("\nSubtype Counts:")
    print(meta_df["Subtype"].value_counts())

    display(meta_df.head(10))

print("\nAll metadata extracted successfully ✅")
```

```python
# =========================================================
# STEP 4.01 — Clean Metadata and Align Samples
# Ensure metadata matches expression matrices
# =========================================================

clean_metadata = {}
clean_matrices = {}

for geo_id in log2_matrices.keys():

    print("="*70)
    print(f"Cleaning {geo_id} ...")

    # Expression matrix
    expr = log2_matrices[geo_id]

    # Metadata
    meta = metadata_dict[geo_id]

    # Remove unknown samples
    meta = meta[meta["Group"] != "Unknown"]

    # Keep only matching samples
    common_samples = list(
        set(expr.columns) &
        set(meta["Sample"])
    )

    # Filter expression matrix
    expr = expr[common_samples]

    # Filter metadata
    meta = meta[
        meta["Sample"].isin(common_samples)
    ]

    # Sort metadata same as matrix columns
    meta = meta.set_index("Sample")
    meta = meta.loc[expr.columns]
    meta = meta.reset_index()

    # Save cleaned data
    clean_metadata[geo_id] = meta
    clean_matrices[geo_id] = expr

    # Summary
    print(f"{geo_id} Alignment Complete ✅")
    print("Expression Shape:", expr.shape)
    print("Metadata Shape:", meta.shape)

    print("\nGroup Counts:")
    print(meta["Group"].value_counts())

    display(meta.head())

print("\nAll datasets cleaned and aligned successfully ✅")
```

```python
# =========================================================
# STEP 4.02 — Normalization and Quality Control (QC)
# =========================================================

import matplotlib.pyplot as plt
from sklearn.decomposition import PCA
from sklearn.preprocessing import StandardScaler

# Store normalized matrices
normalized_matrices = {}

for geo_id, expr in clean_matrices.items():

    print("="*70)
    print(f"Processing QC for {geo_id} ...")

    # -----------------------------------------------------
    # Z-score normalization
    # -----------------------------------------------------
    expr_norm = pd.DataFrame(
        StandardScaler().fit_transform(expr.T).T,
        index=expr.index,
        columns=expr.columns
    )

    normalized_matrices[geo_id] = expr_norm

    print("Normalization Complete ✅")
    print("Shape:", expr_norm.shape)

    # -----------------------------------------------------
    # Boxplot QC
    # -----------------------------------------------------
    plt.figure(figsize=(14,5))

    plt.boxplot(
        expr_norm.values,
        showfliers=False
    )

    plt.title(f"{geo_id} - Boxplot After Normalization")
    plt.xlabel("Samples")
    plt.ylabel("Expression")

    plt.xticks([], [])

    plt.show()

    # -----------------------------------------------------
    # PCA QC
    # -----------------------------------------------------
    pca = PCA(n_components=2)

    pca_result = pca.fit_transform(expr_norm.T)

    meta = clean_metadata[geo_id]

    colors = meta["Group"].map({
        "Cancer": "red",
        "Normal": "blue"
    })

    plt.figure(figsize=(7,6))

    plt.scatter(
        pca_result[:,0],
        pca_result[:,1],
        c=colors
    )

    plt.title(f"{geo_id} - PCA Plot")

    plt.xlabel(
        f"PC1 ({pca.explained_variance_ratio_[0]*100:.2f}% Variance)"
    )

    plt.ylabel(
        f"PC2 ({pca.explained_variance_ratio_[1]*100:.2f}% Variance)"
    )

    plt.show()

print("\nAll datasets normalized and QC checked successfully ✅")


#  CANCER = RED ;   NORMAL = BLUE;
```

```python
# =========================================================
# STEP 4.03 — Batch Effect Correction + Metadata Fix
# Research-Grade Multi-Dataset Integration
# =========================================================

!pip install pycombat -q

import pandas as pd
import numpy as np
from pycombat import Combat

# ---------------------------------------------------------
# STEP 0 — FIX MISCLASSIFIED METADATA (STRICT + PRECISE)
# ---------------------------------------------------------

fixed_metadata = {}

for geo_id, meta in clean_metadata.items():

    meta = meta.copy()

    corrected_groups = []

    for _, row in meta.iterrows():

        text = f"{row['Title']}".lower().strip()

        # -----------------------------------------------
        # NORMAL KEYWORDS
        # -----------------------------------------------
        normal_keywords = [
            "normal",
            "healthy",
            "control",
            "reduction mammoplasty",
            "adjacent normal",
            "non-tumor",
            "non tumor",
            "nontumor",
            "benign"
        ]

        # -----------------------------------------------
        # EXPLICIT CANCER KEYWORDS
        # -----------------------------------------------
        cancer_keywords = [
            "tumor",
            "tumour",
            "cancer",
            "carcinoma",
            "invasive",
            "malignant",
            "metastatic"
        ]

        # -----------------------------------------------
        # STRICT SUBTYPE KEYWORDS (High Precision Only)
        # -----------------------------------------------
        subtype_keywords = [
            "tnbc",
            "triple negative",
            "triple-negative",
            "her2",
            "luminal",
            "basal",
            "claudin-low",
            "claudin low",
            "ductal carcinoma",
            "lobular carcinoma",
            "idc",
            "ilc"
        ]

        # -----------------------------------------------
        # Priority Logic:
        # 1. Explicit cancer/tumor   → Cancer
        # 2. Strict subtype keyword  → Cancer
        # 3. Normal keyword          → Normal
        # 4. Otherwise               → Unknown (dropped)
        # -----------------------------------------------

        if any(k in text for k in cancer_keywords):
            group = "Cancer"

        elif any(k in text for k in subtype_keywords):
            group = "Cancer"

        elif any(k in text for k in normal_keywords):
            group = "Normal"

        else:
            group = "Unknown"

        corrected_groups.append(group)

    meta["Group"] = corrected_groups

    # -----------------------------------------------
    # Report Unknown samples before dropping
    # -----------------------------------------------
    unknown_count = (meta["Group"] == "Unknown").sum()

    if unknown_count > 0:
        print(f"\n⚠ {geo_id}: {unknown_count} Unknown samples detected:")
        print(meta[meta["Group"] == "Unknown"][["index", "Title"]].to_string())

    # Drop unknown
    meta = meta[meta["Group"] != "Unknown"]

    fixed_metadata[geo_id] = meta

    print("=" * 70)
    print(f"{geo_id} Metadata Fixed ✅")
    print(meta["Group"].value_counts())

# ---------------------------------------------------------
# STEP 1 — Find common genes across all datasets
# ---------------------------------------------------------

common_genes = set.intersection(
    *[set(df.index) for df in normalized_matrices.values()]
)

print("\nCommon Genes Found ✅")
print("Total Common Genes:", len(common_genes))

# ---------------------------------------------------------
# STEP 2 — Subset datasets to common genes
# ---------------------------------------------------------

subset_data = {}

for geo_id, df in normalized_matrices.items():

    subset = df.loc[list(common_genes)]

    valid_samples = fixed_metadata[geo_id]["index"].tolist()

    subset = subset[subset.columns.intersection(valid_samples)]

    subset_data[geo_id] = subset

    print(f"{geo_id} subset shape:", subset.shape)

# ---------------------------------------------------------
# STEP 3 — Combine all datasets
# ---------------------------------------------------------

combined_expr = pd.concat(
    subset_data.values(),
    axis=1
)

print("\nCombined Matrix Shape ✅")
print(combined_expr.shape)

# ---------------------------------------------------------
# STEP 4 — Create batch labels
# ---------------------------------------------------------

batch_labels = []

for geo_id, df in subset_data.items():
    batch_labels.extend([geo_id] * df.shape[1])

batch_labels = np.array(batch_labels)

print("\nBatch Labels Created ✅")
print(pd.Series(batch_labels).value_counts())

# ---------------------------------------------------------
# STEP 5 — Combat Batch Correction
# ---------------------------------------------------------

data_array = combined_expr.T.values

combat = Combat()

corrected_data = combat.fit_transform(
    data_array,
    batch_labels
)

corrected_expr = pd.DataFrame(
    corrected_data.T,
    index=combined_expr.index,
    columns=combined_expr.columns
)

print("\nBatch Correction Complete ✅")
print("Corrected Shape:", corrected_expr.shape)

# ---------------------------------------------------------
# STEP 6 — Combine Metadata
# ---------------------------------------------------------

combined_metadata = pd.concat(
    fixed_metadata.values(),
    ignore_index=True
)

combined_metadata = combined_metadata.set_index("index")
combined_metadata = combined_metadata.loc[corrected_expr.columns]
combined_metadata = combined_metadata.reset_index()

print("\nCombined Metadata Shape ✅")
print(combined_metadata.shape)
print("\nGroup Distribution (Final):")
print(combined_metadata["Group"].value_counts())

display(combined_metadata.head(10))

# ---------------------------------------------------------
# FINAL OUTPUT
# ---------------------------------------------------------

print("\nMulti-dataset integration completed successfully ✅")
```

```python
# =========================================================
# STEP 5 — Differential Expression Analysis (DEG)
# limma-style Research-Grade DEG Pipeline
# =========================================================

import pandas as pd
import numpy as np
from scipy import stats
from statsmodels.stats.multitest import multipletests

# ---------------------------------------------------------
# STEP 1 — Prepare metadata
# ---------------------------------------------------------

meta = combined_metadata.copy()

# Use sample names as index
meta = meta.set_index("index")

# ---------------------------------------------------------
# STEP 2 — Extract Cancer and Normal samples
# ---------------------------------------------------------

cancer_samples = meta[
    meta["Group"] == "Cancer"
].index.tolist()

normal_samples = meta[
    meta["Group"] == "Normal"
].index.tolist()

print("Cancer Samples:", len(cancer_samples))
print("Normal Samples:", len(normal_samples))

# ---------------------------------------------------------
# STEP 3 — Subset corrected expression matrix
# ---------------------------------------------------------

expr_cancer = corrected_expr[cancer_samples]

expr_normal = corrected_expr[normal_samples]

# ---------------------------------------------------------
# STEP 4 — DEG Analysis
# ---------------------------------------------------------

results = []

for gene in corrected_expr.index:

    cancer_exp = expr_cancer.loc[gene].dropna()

    normal_exp = expr_normal.loc[gene].dropna()

    # Skip low sample genes
    if len(cancer_exp) < 2 or len(normal_exp) < 2:
        continue

    # Welch's t-test
    t_stat, p_val = stats.ttest_ind(
        cancer_exp,
        normal_exp,
        equal_var=False
    )

    # log2 Fold Change
    logfc = (
        np.mean(cancer_exp) -
        np.mean(normal_exp)
    )

    results.append([
        gene,
        logfc,
        p_val
    ])

# ---------------------------------------------------------
# STEP 5 — Create DEG dataframe
# ---------------------------------------------------------

deg = pd.DataFrame(
    results,
    columns=[
        "Gene",
        "logFC",
        "P.Value"
    ]
)

# ---------------------------------------------------------
# STEP 6 — Remove invalid values
# ---------------------------------------------------------

deg = deg.replace(
    [np.inf, -np.inf],
    np.nan
)

deg = deg.dropna()

deg = deg[
    (deg["P.Value"] >= 0) &
    (deg["P.Value"] <= 1)
]

# ---------------------------------------------------------
# STEP 7 — Multiple testing correction (FDR)
# ---------------------------------------------------------

deg["adj.P.Val"] = multipletests(
    deg["P.Value"],
    method="fdr_bh"
)[1]

# ---------------------------------------------------------
# STEP 8 — DEG filtering
# Paper-style thresholds
# ---------------------------------------------------------

significant_deg = deg[
    (abs(deg["logFC"]) > 1) &
    (deg["adj.P.Val"] < 0.05)
].copy()

# Sort by significance
significant_deg = significant_deg.sort_values(
    "adj.P.Val"
)

# ---------------------------------------------------------
# STEP 9 — Upregulated / Downregulated
# ---------------------------------------------------------

upregulated = significant_deg[
    significant_deg["logFC"] > 1
]

downregulated = significant_deg[
    significant_deg["logFC"] < -1
]

# ---------------------------------------------------------
# FINAL SUMMARY
# ---------------------------------------------------------

print("\nDEG Analysis Complete ✅")

print("\nTotal Significant DEGs:",
      len(significant_deg))

print("Upregulated Genes:",
      len(upregulated))

print("Downregulated Genes:",
      len(downregulated))

print("\nTop Significant DEGs:")

display(significant_deg.head(20))
```

```python
# =========================================================
# STEP 8.03 — lncRNA–mRNA Co-expression Analysis
# Paper-Style Co-expression Network
# =========================================================

import pandas as pd
import numpy as np
from scipy.stats import pearsonr

# ---------------------------------------------------------
# STEP 1 — Prepare Gene Lists
# ---------------------------------------------------------

lncrnas = lncrna_deg["Gene"].unique().tolist()

mrnas = [
    g for g in significant_deg["Gene"].unique()
    if g not in lncrnas
]

print("lncRNAs:", len(lncrnas))
print("mRNAs:", len(mrnas))

# ---------------------------------------------------------
# STEP 2 — Correlation Analysis
# ---------------------------------------------------------

results = []

for lnc in lncrnas:

    if lnc not in corrected_expr.index:
        continue

    lnc_exp = corrected_expr.loc[lnc]

    for gene in mrnas:

        if gene not in corrected_expr.index:
            continue

        gene_exp = corrected_expr.loc[gene]

        try:

            r, p = pearsonr(
                lnc_exp,
                gene_exp
            )

            results.append([
                lnc,
                gene,
                r,
                p
            ])

        except:
            pass

# ---------------------------------------------------------
# STEP 3 — Create Table
# ---------------------------------------------------------

coexp = pd.DataFrame(
    results,
    columns=[
        "lncRNA",
        "mRNA",
        "Correlation",
        "P.Value"
    ]
)

# ---------------------------------------------------------
# STEP 4 — Paper-style Filtering
# ---------------------------------------------------------

coexp = coexp[
    (abs(coexp["Correlation"]) >= 0.60) &
    (coexp["P.Value"] < 0.05)
].copy()

# ---------------------------------------------------------
# STEP 5 — Rank Strongest Pairs
# ---------------------------------------------------------

coexp["AbsCorr"] = abs(
    coexp["Correlation"]
)

coexp = coexp.sort_values(
    "AbsCorr",
    ascending=False
)

print(
    "\nSignificant lncRNA–mRNA Pairs Found ✅"
)

print("Total Pairs:", len(coexp))

display(
    coexp.head(30)
)

# ---------------------------------------------------------
# STEP 6 — Export
# ---------------------------------------------------------

coexp.to_csv(
    "lncRNA_mRNA_Coexpression.csv",
    index=False
)

print(
    "\nExported: lncRNA_mRNA_Coexpression.csv ✅"
)
```

```python
# =========================================================
# STEP 10.02 — STRING PPI Export + Hub Gene Analysis
# Colab-Based Cytoscape Alternative
# =========================================================


# Cytoscape alt = Python-based centrality + hub analysis;


import pandas as pd
import networkx as nx
import matplotlib.pyplot as plt
import numpy as np

# ---------------------------------------------------------
# STEP 1 — Export STRING Network for Cytoscape
# ---------------------------------------------------------

cytoscape_edges = ppi[[
    "preferredName_A",
    "preferredName_B",
    "score"
]].copy()

cytoscape_edges.columns = [
    "Source",
    "Target",
    "ConfidenceScore"
]

# Remove duplicate interactions
cytoscape_edges = cytoscape_edges.drop_duplicates()

# Save edge table
cytoscape_edges.to_csv(
    "Cytoscape_STRING_Edges.csv",
    index=False
)

print("STRING Edge File Exported ✅")
print("File: Cytoscape_STRING_Edges.csv")

display(cytoscape_edges.head())

# ---------------------------------------------------------
# STEP 2 — Build PPI Network
# ---------------------------------------------------------

G = nx.Graph()

for _, row in ppi.iterrows():

    G.add_edge(
        row["preferredName_A"],
        row["preferredName_B"],
        weight=row["score"]
    )

print("\nPPI Network Constructed ✅")

print("Nodes:", G.number_of_nodes())
print("Edges:", G.number_of_edges())

# ---------------------------------------------------------
# STEP 3 — Calculate Network Centrality
# Paper-style Hub Approximation
# ---------------------------------------------------------

degree_centrality = nx.degree_centrality(G)

betweenness_centrality = nx.betweenness_centrality(G)

closeness_centrality = nx.closeness_centrality(G)

eigenvector_centrality = nx.eigenvector_centrality(
    G,
    max_iter=1000
)

# ---------------------------------------------------------
# STEP 4 — Create Hub Gene Table
# ---------------------------------------------------------

hub_df = pd.DataFrame({

    "Gene": list(G.nodes()),

    "DegreeCentrality": [
        degree_centrality[g]
        for g in G.nodes()
    ],

    "BetweennessCentrality": [
        betweenness_centrality[g]
        for g in G.nodes()
    ],

    "ClosenessCentrality": [
        closeness_centrality[g]
        for g in G.nodes()
    ],

    "EigenvectorCentrality": [
        eigenvector_centrality[g]
        for g in G.nodes()
    ]
})

# Combined score
hub_df["HubScore"] = (

    hub_df["DegreeCentrality"] +

    hub_df["BetweennessCentrality"] +

    hub_df["ClosenessCentrality"] +

    hub_df["EigenvectorCentrality"]

)

# Sort hubs
hub_df = hub_df.sort_values(
    by="HubScore",
    ascending=False
).reset_index(drop=True)

print("\nTop Hub Genes Identified ✅")

display(hub_df.head(20))

# ---------------------------------------------------------
# STEP 5 — Export Hub Gene Table
# ---------------------------------------------------------

hub_df.to_csv(
    "Hub_Genes_Table.csv",
    index=False
)

print("\nHub Gene Table Exported ✅")
print("File: Hub_Genes_Table.csv")

# ---------------------------------------------------------
# STEP 6 — Create Cytoscape Node Table
# ---------------------------------------------------------

node_table = hub_df.merge(

    significant_deg[[
        "Gene",
        "logFC",
        "adj.P.Val"
    ]],

    on="Gene",
    how="left"
)

# Regulation label
node_table["Regulation"] = np.where(
    node_table["logFC"] > 1,
    "Upregulated",
    "Downregulated"
)

# Export node table
node_table.to_csv(
    "Cytoscape_Node_Table.csv",
    index=False
)

print("\nCytoscape Node Table Exported ✅")
print("File: Cytoscape_Node_Table.csv")

display(node_table.head())

# ---------------------------------------------------------
# STEP 7 — Visualize Top Hub Genes
# ---------------------------------------------------------

top_hubs = hub_df.head(15)

plt.figure(figsize=(10,6))

plt.barh(
    top_hubs["Gene"],
    top_hubs["HubScore"]
)

plt.xlabel("Combined Hub Score")

plt.title(
    "Top Hub Genes"
)

plt.gca().invert_yaxis()

plt.show()

# ---------------------------------------------------------
# STEP 8 — Highlight Hub Network
# ---------------------------------------------------------

top_gene_set = set(
    top_hubs["Gene"]
)

node_colors = [
    "red" if n in top_gene_set else "lightgray"
    for n in G.nodes()
]

node_sizes = [
    2500 if n in top_gene_set else 300
    for n in G.nodes()
]

plt.figure(figsize=(15,12))

pos = nx.spring_layout(
    G,
    seed=42,
    k=0.4
)

nx.draw_networkx_nodes(
    G,
    pos,
    node_color=node_colors,
    node_size=node_sizes,
    alpha=0.9
)

nx.draw_networkx_edges(
    G,
    pos,
    alpha=0.4
)

nx.draw_networkx_labels(
    G,
    pos,
    font_size=8
)

plt.title(
    "Hub Gene Network Visualization"
)

plt.axis("off")

plt.show()

print("\nSTEP 10.02 Completed ✅")
```

```python
import pandas as pd
import networkx as nx
import matplotlib.pyplot as plt
import numpy as np
import math

# 1. Analyze the FULL Graph (G) for MCC to avoid the '2-node' bottleneck
# CytoHubba MCC = ∑ (size of clique - 1)!
mcc_scores = {}
for node in G.nodes():
    score = 0
    # Find cliques in the neighborhood of the node
    cliques = [c for c in nx.find_cliques(G) if node in c]
    for c in cliques:
        score += math.factorial(len(c) - 1)
    mcc_scores[node] = score

# 2. Build Ranking Table
mcc_df = pd.DataFrame({
    "Gene": list(G.nodes()),
    "MCC_Score": [mcc_scores[g] for g in G.nodes()],
    "DegreeCentrality": [nx.degree_centrality(G)[g] for g in G.nodes()],
    "BetweennessCentrality": [nx.betweenness_centrality(G)[g] for g in G.nodes()]
})

mcc_df = mcc_df.sort_values("MCC_Score", ascending=False).reset_index(drop=True)
print("\nTop MCC Hub Genes (Calculated from Full Network) ✅")
display(mcc_df.head(15))

# 3. Plot
plt.figure(figsize=(10,6))
plt.barh(mcc_df.head(15)["Gene"], mcc_df.head(15)["MCC_Score"])
plt.title("CytoHubba-Style MCC Ranking (Top 15)")
plt.gca().invert_yaxis()
plt.show()

final_hub_genes = mcc_df.head(10)["Gene"].tolist()
print(f"Final Hub Candidates identified: {len(final_hub_genes)}")
```

```python
# =========================================================
# STEP 12 — Survival Analysis of Final MCC Hub Genes
# =========================================================

!pip install lifelines -q

import pandas as pd
import numpy as np
import matplotlib.pyplot as plt

from lifelines import KaplanMeierFitter
from lifelines.statistics import logrank_test

# ---------------------------------------------------------
# STEP 1 — Select Final MCC Hub Genes
# ---------------------------------------------------------

survival_genes = final_hub_table[
    "Gene"
].head(5).tolist()

print("Genes Selected for Survival Analysis ✅")

print(survival_genes)

# ---------------------------------------------------------
# STEP 2 — Survival Data Notice
# ---------------------------------------------------------

print("""
WARNING ⚠️

Current analysis uses simulated survival data.

For publication-grade survival validation:
- TCGA-BRCA Clinical Data
or
- KM Plotter
should be used.
""")

# ---------------------------------------------------------
# STEP 3 — Simulated Clinical Data
# ---------------------------------------------------------

np.random.seed(42)

clinical_data = pd.DataFrame({

    "Sample": corrected_expr.columns,

    "Overall_Survival": np.random.randint(
        200,
        4000,
        size=corrected_expr.shape[1]
    ),

    "Event": np.random.binomial(
        1,
        0.45,
        size=corrected_expr.shape[1]
    )
})

print("\nClinical Table Created ✅")

display(
    clinical_data.head()
)

# ---------------------------------------------------------
# STEP 4 — Survival Summary Table
# ---------------------------------------------------------

survival_results = []

# ---------------------------------------------------------
# STEP 5 — Kaplan-Meier Analysis
# ---------------------------------------------------------

for gene in survival_genes:

    print("\n" + "="*70)

    print(f"Survival Analysis: {gene}")

    # ---------------------------------------------
    # Expression Values
    # ---------------------------------------------

    if gene not in corrected_expr.index:

        print("Gene not found ❌")

        continue

    gene_exp = corrected_expr.loc[gene]

    median_exp = gene_exp.median()

    high_group = gene_exp[
        gene_exp >= median_exp
    ].index

    low_group = gene_exp[
        gene_exp < median_exp
    ].index

    # ---------------------------------------------
    # Clinical Merge
    # ---------------------------------------------

    high_data = clinical_data[
        clinical_data["Sample"].isin(
            high_group
        )
    ]

    low_data = clinical_data[
        clinical_data["Sample"].isin(
            low_group
        )
    ]

    # ---------------------------------------------
    # KM Fit
    # ---------------------------------------------

    kmf_high = KaplanMeierFitter()

    kmf_low = KaplanMeierFitter()

    plt.figure(figsize=(8,6))

    kmf_high.fit(

        durations=high_data[
            "Overall_Survival"
        ],

        event_observed=high_data[
            "Event"
        ],

        label=f"{gene} High"
    )

    kmf_high.plot_survival_function()

    kmf_low.fit(

        durations=low_data[
            "Overall_Survival"
        ],

        event_observed=low_data[
            "Event"
        ],

        label=f"{gene} Low"
    )

    kmf_low.plot_survival_function()

    # ---------------------------------------------
    # Log-rank Test
    # ---------------------------------------------

    result = logrank_test(

        high_data[
            "Overall_Survival"
        ],

        low_data[
            "Overall_Survival"
        ],

        event_observed_A=
        high_data["Event"],

        event_observed_B=
        low_data["Event"]
    )

    p_value = result.p_value

    survival_results.append([
        gene,
        p_value
    ])

    plt.title(
        f"{gene}\n"
        f"Log-rank p = {p_value:.4e}"
    )

    plt.xlabel(
        "Survival Time"
    )

    plt.ylabel(
        "Survival Probability"
    )

    plt.show()

# ---------------------------------------------------------
# STEP 6 — Survival Ranking Table
# ---------------------------------------------------------

survival_df = pd.DataFrame(

    survival_results,

    columns=[
        "Gene",
        "P.Value"
    ]
)

survival_df = survival_df.sort_values(
    "P.Value"
)

print("\nSurvival Ranking Table ✅")

display(
    survival_df
)

# ---------------------------------------------------------
# STEP 7 — Significant Survival Genes
# ---------------------------------------------------------

significant_survival = survival_df[

    survival_df[
        "P.Value"
    ] < 0.05

].copy()

print(
    "\nSignificant Survival Genes ✅"
)

display(
    significant_survival
)

# ---------------------------------------------------------
# STEP 8 — Export
# ---------------------------------------------------------

survival_df.to_csv(
    "Survival_Analysis_Results.csv",
    index=False
)

print(
    "\nExported: Survival_Analysis_Results.csv ✅"
)

print(
    "\nSTEP 12 Completed Successfully ✅"
)
```

```python
# =========================================================
# STEP 13 — Expression Validation of Final MCC Hub Genes
# =========================================================

import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns

from scipy.stats import ttest_ind

# ---------------------------------------------------------
# STEP 1 — Select Final Hub Genes
# ---------------------------------------------------------

validation_genes = final_hub_table[
    "Gene"
].head(10).tolist()

print("Genes Selected for Validation ✅")

print(validation_genes)

# ---------------------------------------------------------
# STEP 2 — Prepare Metadata
# ---------------------------------------------------------

meta = combined_metadata.copy()

meta = meta.set_index("index")

cancer_samples = meta[
    meta["Group"] == "Cancer"
].index.tolist()

normal_samples = meta[
    meta["Group"] == "Normal"
].index.tolist()

print("\nCancer Samples:", len(cancer_samples))
print("Normal Samples:", len(normal_samples))

# ---------------------------------------------------------
# STEP 3 — Expression Validation
# ---------------------------------------------------------

validation_results = []

for gene in validation_genes:

    if gene not in corrected_expr.index:
        continue

    cancer_exp = corrected_expr.loc[
        gene,
        cancer_samples
    ]

    normal_exp = corrected_expr.loc[
        gene,
        normal_samples
    ]

    stat, pval = ttest_ind(
        cancer_exp,
        normal_exp,
        equal_var=False
    )

    validation_results.append([

        gene,

        np.mean(cancer_exp),

        np.mean(normal_exp),

        np.mean(cancer_exp) -
        np.mean(normal_exp),

        pval
    ])

    # -----------------------------------------------------
    # Boxplot
    # -----------------------------------------------------

    plot_df = pd.DataFrame({

        "Expression": np.concatenate([
            cancer_exp.values,
            normal_exp.values
        ]),

        "Group": (
            ["Cancer"] * len(cancer_exp)
            +
            ["Normal"] * len(normal_exp)
        )
    })

    plt.figure(figsize=(6,5))

    sns.boxplot(
        x="Group",
        y="Expression",
        data=plot_df
    )

    sns.stripplot(
        x="Group",
        y="Expression",
        data=plot_df,
        alpha=0.45
    )

    plt.title(
        f"{gene}\n"
        f"P = {pval:.2e}"
    )

    plt.show()

# ---------------------------------------------------------
# STEP 4 — Validation Table
# ---------------------------------------------------------

validation_df = pd.DataFrame(

    validation_results,

    columns=[

        "Gene",

        "CancerMean",

        "NormalMean",

        "MeanDifference",

        "P.Value"
    ]
)

validation_df = validation_df.sort_values(
    "P.Value"
)

print("\nExpression Validation Complete ✅")

display(validation_df)

# ---------------------------------------------------------
# STEP 5 — Significant Genes
# ---------------------------------------------------------

significant_validation = validation_df[
    validation_df["P.Value"] < 0.05
]

print(
    "\nSignificant Validated Genes ✅"
)

display(
    significant_validation
)

# ---------------------------------------------------------
# STEP 6 — Export
# ---------------------------------------------------------

validation_df.to_csv(
    "Expression_Validation_Results.csv",
    index=False
)

print(
    "\nExported: Expression_Validation_Results.csv ✅"
)

print(
    "\nSTEP 13 Completed Successfully ✅"
)
```

```python
import pandas as pd
from sklearn.model_selection import train_test_split
from sklearn.preprocessing import StandardScaler

# Features (X) are the corrected expression values
X = corrected_expr.T # Transpose to have samples as rows and genes as columns

# Target (y) is the 'Group' from combined_metadata
# Ensure the order of samples matches X
y = combined_metadata.set_index('index').loc[X.index]['Group']

# Convert 'Cancer' to 1 and 'Normal' to 0
y = y.map({'Cancer': 1, 'Normal': 0})

# Split data into training and testing sets
X_train, X_test, y_train, y_test = train_test_split(
    X,
    y,
    test_size=0.3,
    random_state=42,
    stratify=y # Maintain class balance
)

# Scale the features
scaler = StandardScaler()
X_train_scaled = scaler.fit_transform(X_train)
X_test_scaled = scaler.transform(X_test)

print("Data preparation complete ✅")
print(f"X_train shape: {X_train_scaled.shape}")
print(f"y_train shape: {y_train.shape}")
print(f"X_test shape: {X_test_scaled.shape}")
print(f"y_test shape: {y_test.shape}")
```

```python
from sklearn.linear_model import LassoCV
import numpy as np

# Initialize LassoCV to find the optimal alpha using cross-validation
# The 'max_iter' is increased to ensure convergence for potentially complex datasets
lasso_model = LassoCV(
    cv=5,
    random_state=42,
    max_iter=10000,
    n_jobs=-1 # Use all available cores for parallel processing
).fit(X_train_scaled, y_train)

print(f"Optimal alpha found by LassoCV: {lasso_model.alpha_}")

# Get the coefficients
lasso_coefficients = pd.Series(lasso_model.coef_, index=X.columns)

# Identify non-zero coefficients (selected features)
selected_features = lasso_coefficients[lasso_coefficients != 0].sort_values(ascending=False)

print(f"Number of features selected by Lasso: {len(selected_features)}")
print("\nTop 10 selected features (genes) and their coefficients:")
display(selected_features.head(10))
```

```python
from sklearn.metrics import accuracy_score, classification_report, roc_auc_score, roc_curve
import matplotlib.pyplot as plt
import seaborn as sns

# Predict on the test set
y_pred = lasso_model.predict(X_test_scaled)

# Convert probabilities to binary predictions (0 or 1)
y_pred_binary = (y_pred > 0.5).astype(int)

print(f"Accuracy: {accuracy_score(y_test, y_pred_binary):.4f}")
print("\nClassification Report:")
print(classification_report(y_test, y_pred_binary))

# ROC AUC Score
roc_auc = roc_auc_score(y_test, y_pred)
print(f"ROC AUC Score: {roc_auc:.4f}")

# Plot ROC curve
fpr, tpr, _ = roc_curve(y_test, y_pred)

plt.figure(figsize=(8, 6))
plt.plot(fpr, tpr, color='darkorange', lw=2, label=f'ROC curve (AUC = {roc_auc:.2f})')
plt.plot([0, 1], [0, 1], color='navy', lw=2, linestyle='--')
plt.xlabel('False Positive Rate')
plt.ylabel('True Positive Rate')
plt.title('Receiver Operating Characteristic (ROC) Curve')
plt.legend(loc='lower right')
plt.grid(True)
plt.show()

print("Lasso model evaluation complete ✅")
```
